```python

# Step 1: Import libraries
from transformers import pipeline
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt

# Step 2: Load the object detection model
object_detector = pipeline(
    "object-detection",
    model="facebook/detr-resnet-50"
)

# Step 3: Load the traffic image
traffic_img = Image.open("/content/Untitled.png").convert("RGB")

# Display original image
plt.figure(figsize=(10, 8))
plt.imshow(traffic_img)
plt.axis("off")
plt.show()

# Step 4: Pass the input to the pipeline
detections = object_detector(traffic_img)

# Display first 4 detections
print(detections[:4])

# Step 5: Create a drawing context
draw = ImageDraw.Draw(traffic_img)

# Step 6: Define a font
try:
    font = ImageFont.truetype("DejaVuSans.ttf", 20)
except OSError:
    font = ImageFont.load_default()

# Step 7: Draw bounding boxes and labels
for detection in detections:

    box = detection['box']
    label = detection['label']
    score = detection['score']

    # Define rectangle coordinates
    xmin = int(box['xmin'])
    ymin = int(box['ymin'])
    xmax = int(box['xmax'])
    ymax = int(box['ymax'])

    # Draw the rectangle
    draw.rectangle(
        [(xmin, ymin), (xmax, ymax)],
        outline="red",
        width=4
    )

    # Prepare label text
    text = f"{label}: {score:.2f}"

    # Get text size
    text_bbox = draw.textbbox((xmin, ymin), text, font=font)

    text_width = text_bbox[2] - text_bbox[0]
    text_height = text_bbox[3] - text_bbox[1]

    # Calculate label position
    label_y = max(0, ymin - text_height - 5)

    # Draw text background rectangle
    draw.rectangle(
        [
            (xmin, label_y),
            (xmin + text_width + 6, label_y + text_height + 6)
        ],
        fill="red"
    )

    # Draw the text
    draw.text(
        (xmin + 3, label_y + 3),
        text,
        fill="white",
        font=font
    )

# Step 8: Display the final image
plt.figure(figsize=(12, 8))
plt.imshow(traffic_img)
plt.axis("off")
plt.show()
```
